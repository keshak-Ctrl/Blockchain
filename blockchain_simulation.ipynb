```python
#------------------------------------------------------------------------------
# Strucutre of Block
#------------------------------------------------------------------------------
import hashlib
import time

class Block:
    def __init__(self, index, timestamp, data, previous_hash):
        self.index = index
        self.timestamp = timestamp
        self.data = data
        self.previous_hash = previous_hash
        self.nonce = 0
        self.hash = self.calculate_hash()

    def calculate_hash(self):
        block_string = f"{self.index}{self.timestamp}{self.data}{self.previous_hash}"
        return hashlib.sha256(block_string.encode()).hexdigest()
```

```python
#------------------------------------------------------------------------------
# minning the block
#------------------------------------------------------------------------------
def mine_block(block, difficulty=3, verbose=False):
    """
    Mines a block by finding a nonce such that the block's hash starts with 'difficulty' zeros.
    Prints detailed debug information on each iteration if verbose is True.
    """
    target = '0' * difficulty
    while block.hash[:difficulty] != target:
        block.nonce += 1
        block.hash = block.calculate_hash()
        if verbose and block.nonce % 100000 == 0:
            print(f"Mining... Nonce: {block.nonce}, Hash: {block.hash}")
    return block
```
